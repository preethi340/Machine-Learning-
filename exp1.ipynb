```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


def eda_process(filepath, target_column=None):

    
    print(f"LOADING: {filepath}")
   

    df = pd.read_csv(filepath)

    print(f"Shape: {df.shape}")
    print(df.head())

    print("\nData types:\n", df.dtypes)

    
    print("SUMMARY STATISTICS")
    

    print(df.describe())

    print("\nMissing values (NaN):")
    print(df.isnull().sum())

    print("\nDuplicate rows:", df.duplicated().sum())

    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

    if target_column and target_column in numeric_cols:
        numeric_cols.remove(target_column)

    
    print("VISUALIZATIONS")
   

    plot_cols = numeric_cols[:10]

    df[plot_cols].hist(figsize=(12,8))
    plt.suptitle("Feature Distributions")
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(10,8))
    sns.heatmap(df[plot_cols].corr(), annot=True, cmap="coolwarm")
    plt.title("Correlation Heatmap")
    plt.show()

    if len(numeric_cols) <= 6:
        if target_column:
            sns.pairplot(df,
                         vars=numeric_cols,
                         hue=target_column)
        else:
            sns.pairplot(df[numeric_cols])

        plt.suptitle("Pair Plot", y=1.02)
        plt.show()

    else:
        print(f"Skipped Pairplot : {len(numeric_cols)} numeric columns.")

    return df

df_iris = eda_process(
    filepath="Iris.csv",
    target_column="Species"
)

df_loan = eda_process(
    filepath="loan.csv",
    target_column="LoanAmount"
)

df_diabetes = eda_process(
    filepath="diabetes.csv",
    target_column="Outcome"
)

df_spam = eda_process(
    filepath="emails.csv",
    target_column="Prediction"
)

df_sample = pd.read_csv("mnist.csv", nrows=2000)

df_sample.to_csv("mnist_sample.csv", index=False)

df_mnist = eda_process(
    filepath="mnist_sample.csv",
    target_column="label"
)
```

```python
from sklearn.preprocessing import StandardScaler


def detect_task_type(df, target_column, unique_threshold=20):

    target_dtype = df[target_column].dtype
    n_unique = df[target_column].nunique()

    print("\nTarget column:", target_column)
    print("Datatype:", target_dtype)
    print("Unique values:", n_unique)

    if target_dtype == "object" or target_dtype.name == "category":
        task_type = "Classification"

    elif n_unique <= unique_threshold:
        task_type = "Classification"

    else:
        task_type = "Regression"

    print("Detected Task :", task_type)

    return task_type


def preprocessing_process(df, target_column, zero_as_missing_cols=None):

 
    print("DATA PREPROCESSING")
  

    task_type = detect_task_type(df, target_column)

    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

    if target_column in numeric_cols:
        numeric_cols.remove(target_column)


    print("\nHandling Missing Values...")

    if zero_as_missing_cols:
        df[zero_as_missing_cols] = df[zero_as_missing_cols].replace(0, np.nan)

        print("Converted 0 -> NaN in:", zero_as_missing_cols)

    for col in numeric_cols:

        missing = df[col].isnull().sum()

        if missing > 0:

            df[col] = df[col].fillna(df[col].median())

            print(f"{col} : {missing} values filled with median")

    rows_before = df.shape[0]

    df.dropna(inplace=True)

    print("Rows removed:", rows_before-df.shape[0])


    print("\nHandling Outliers...")

    outlier_summary = {}

    for col in numeric_cols:

        Q1 = df[col].quantile(0.25)

        Q3 = df[col].quantile(0.75)

        IQR = Q3-Q1

        lower = Q1-1.5*IQR

        upper = Q3+1.5*IQR

        outliers = df[(df[col]<lower) | (df[col]>upper)]

        outlier_summary[col]=len(outliers)

        df[col]=df[col].clip(lower,upper)

    print("Outliers handled")

    print(outlier_summary)

  

    print("\nNormalizing Features...")

    X = df.drop(columns=[target_column])
    X = X.select_dtypes(include=np.number)

    y = df[target_column]

    scaler = StandardScaler()

    X_scaled = scaler.fit_transform(X)

    X_scaled = pd.DataFrame(X_scaled,
                            columns=X.columns,
                            index=df.index)

    df_processed = pd.concat([X_scaled, y], axis=1)

    print("Normalization Completed")

    return df_processed, task_type

df_iris, task = preprocessing_process(
    df_iris,
    target_column="Species"
)


df_loan, task = preprocessing_process(
    df_loan,
    target_column="LoanAmount"
)

df_diabetes, task = preprocessing_process(
    df_diabetes,
    target_column="Outcome",
    zero_as_missing_cols=[
        'Glucose',
        'BloodPressure',
        'SkinThickness',
        'Insulin',
        'BMI'
    ]
)

df_spam, task = preprocessing_process(
    df_spam,
    target_column="Prediction"
)

df_mnist, task = preprocessing_process(
    df_mnist,
    target_column="label"
)
```
